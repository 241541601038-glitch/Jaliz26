```python
def login(username, password):
    """
    Checks if the provided username and password match expected values.
    In a real app, this would check a secure database.
    """
    valid_username = "jaliz"
    valid_password = "2626"

    if username == valid_username and password == valid_password:
        return True
    else:
        return False

if __name__ == "__main__":
    print("--- Simple Login System ---")
    
    user_input = input("Enter your username: ")
    import getpass
    password_input = getpass.getpass("Enter your password: ")

    if login(user_input, password_input):
        print("Login successful! Welcome, {}.".format(user_input))
    else:
        print("Invalid username or password.")
```
