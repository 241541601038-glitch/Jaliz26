```python
print("Enter Marks Obtained in 5 Subjects (out of 100 each):")
subject1 = float(input("Python marks: "))
subject2 = float(input("Linux marks: "))
subject3 = float(input("IoT marks: "))
subject4 = float(input("KE marks: "))
subject5 = float(input("ML marks: "))

total_marks = subject1 + subject2 + subject3 + subject4 + subject5
max_marks = 500
percentage = (total_marks / max_marks) * 100

if percentage >= 90:
    grade = "A"
elif percentage >= 80:
    grade = "B"
elif percentage >= 70:
    grade = "C"
elif percentage >= 60:
    grade = "D"
else:
    grade = "F"

print(f"\nTotal Marks: {total_marks:.2f} out of {max_marks}")
print(f"Percentage: {percentage:.2f}%")
print(f"Grade: {grade}")
```
